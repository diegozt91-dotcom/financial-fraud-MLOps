window.toggleZMenu = function(event) {
    event.stopPropagation();
    const btn = event.currentTarget;
    const menu = document.getElementById('z-menu-items');
    
    if (menu.style.display === 'flex') {
        menu.style.display = 'none';
    } else {
        const bounds = btn.getBoundingClientRect();
        menu.style.top = bounds.bottom + 'px';
        menu.style.left = bounds.left + 'px';
        menu.style.display = 'flex';
    }
};

window.closeZMenu = function() {
    const menu = document.getElementById('z-menu-items');
    if (menu) menu.style.display = 'none';
};

Para el desarrollo de las características aplicaremos algunas funciones matemáticas. Nosotros trabajaremos con la distribución de von Mises para la determinación de algunas características temporales, a continuación mostramos las fórmulas para el cálculo de la media periódica (1) y la desviación estándar periódica (2). Para el cálculo de las distancias aplicaremos la fórmula de Haversine (3). 
$$\mu_{vM}(D) = 2 \tan^{-1} \left( \frac{\sum_{t_j \in D} \sin(t_j)}{\left(\sqrt{\left(\sum_{t_j \in D} \cos(t_j)\right)^2 + \left(\sum_{t_j \in D} \sin(t_j)\right)^2} + \sum_{t_j \in D} \cos(t_j)\right)} \right), \tag{1}$$
  
$$\sigma_{vM}(D) = \sqrt{\ln \left( \frac{1}{\left(\frac{1}{N} \sum_{t_j \in D} \sin(t_j)\right)^2 + \left(\frac{1}{N} \sum_{t_j \in D} \cos(t_j)\right)^2} \right)}, \tag{2}$$

$$d = 2R \cdot \text{atan2}\left(\sqrt{\sin^2\left(\frac{\Delta \varphi}{2}\right) + \cos(\varphi_1) \cos(\varphi_2) \sin^2\left(\frac{\Delta \lambda}{2}\right)}, \sqrt{1 - \sin^2\left(\frac{\Delta \varphi}{2}\right) - \cos(\varphi_1) \cos(\varphi_2) \sin^2\left(\frac{\Delta \lambda}{2}\right)}\right). \tag{3}$$

<br>Para trabajar con estas fórmulas, realizar la estadística y algunas verficaciones, definimos algunas funciones a continuación en la siguiente celda. 

````python
## En el artículo referenciado se utiliza la arcotangente, por tanto nosotros debemos trabajar con ángulos 
## que oscilen entre - pi y pi. Definimos una función para la normalización
PI = math.pi
def normalize_angle(angle):
    return ((angle + PI) % (2 * PI)) - PI

## Definimos la función de Haversine con arctan2 para evitar problemas derivados de arcsin en distancias cortas
R_EARTH = 6371.0088
def calculate_haversine(lat1, lon1, lat2, lon2):
    dlat = pl.col(lat2) - pl.col(lat1)
    dlon = pl.col(lon2) - pl.col(lon1)
    a = (dlat / 2).sin().pow(2) + pl.col(lat1).cos() * pl.col(lat2).cos() * (dlon / 2).sin().pow(2)
    return 2 * R_EARTH * pl.arctan2(a.sqrt(), (1 - a).sqrt())

# Calculamos los valores críticos utilizando una distribución normal si la muestra, N_sample, es mayor a 30, 
# en caso contrario utilizaremos una distribucion normal por la teoria del limite central
T_DISTRIBUTIONS = {
    90: {n: stats.t.ppf(0.95, df=n-1) for n in range(2, 31)},
    95: {n: stats.t.ppf(0.975, df=n-1) for n in range(2, 31)},
    99: {n: stats.t.ppf(0.995, df=n-1) for n in range(2, 31)}
}
Z_VALUES = {90: 1.645, 95: 1.960, 99: 2.576}
def calculate_Zt_value(N_sample: str, confidence_level: int) -> pl.Expr:
    if confidence_level not in T_DISTRIBUTIONS:
        raise ValueError("confidence_level must be 90, 95 or 99")
    z_val = Z_VALUES[confidence_level]
    t_dist = T_DISTRIBUTIONS[confidence_level]
    expression = pl.when(pl.col(N_sample) >= 31).then(z_val)    
    for n_val, t_val in t_dist.items():
        expression = expression.when(pl.col(N_sample) == n_val).then(t_val)
    return expression.otherwise(0.0)

## Creamos una función que verifique que el número de filas coincida con el número de ids
def check_row_id(lazyframe):    
    check = lazyframe.select(
        pl.len().alias("rows"),
        pl.col("id").n_unique().alias("unique_ids")
    ).collect()
    return check

## Creamos una función que verifique los valores de las columnas númericas de nuestro lazyframe
def calculate_INF_NULL_NAN(lazyframe):    
    numtype_columns = lazyframe.select(cs.numeric()).collect_schema().names()
    analysis = pl.concat(
        [lazyframe.select(
            pl.lit(col).alias("COLUMN"),
            pl.col(col).is_infinite().sum().alias("INF"),
            pl.col(col).is_null().sum().alias("NULL"),
            pl.col(col).is_nan().sum().alias("NAN"),
         )
         for col in numtype_columns 
        ]
    ).collect()    
    with pl.Config(tbl_rows=-1, fmt_str_lengths=100):
        print(analysis)
````

En esta sección definimos algunas funciones que utilizaremos para evaluar el entrenamiento del modelo

````python
## Funciones para el entrenamiento por machine learning

def create_validation_report(modelo, calibrated_threshold = 0.45):
    val_probs = modelo.predict_proba(X_val)[:, 1]
    calibrated_threshold = calibrated_threshold
    calibrated_preds = val_probs > calibrated_threshold
    print("-------- VALIDATION REPORT --------")
    print(classification_report(y_val, calibrated_preds))    
    precision, recall, thresholds = precision_recall_curve(y_val, val_probs)
    pr_auc = auc(recall, precision)
    ap = average_precision_score(y_val, val_probs)
    print(f"PR-AUC Score: {pr_auc:.4f}")
    print(f"Average Precision (AP) Score: {ap:.4f}")

def create_test_report(modelo, calibrated_threshold = 0.45):
    test_probs = modelo.predict_proba(X_test)[:, 1]
    test_preds = test_probs > calibrated_threshold
    print("----------- TEST REPORT -----------")
    print(classification_report(y_test, test_preds))
    precision, recall, thresholds = precision_recall_curve(y_test, test_probs)
    pr_auc = auc(recall, precision)
    ap = average_precision_score(y_test, test_probs)
    print(f"PR-AUC Score: {pr_auc:.4f}")
    print(f"Average Precision (AP) Score: {ap:.4f}")
    return precision, recall, pr_auc, ap, test_preds

## Funciones para el entrenamiento por deeep learning

def plot_monitoring(history, name):
    df_history = pd.DataFrame(history.history)
    df_history['Epoch'] = df_history.index + 1
    fig_auc = px.line(
        df_history, 
        x='Epoch', 
        y=['pr_auc', 'val_pr_auc'], 
        labels={'value': 'PR-AUC score', 'variable': 'Set'},
        title=f"Monitoring: {name}"
    )
    new_names_auc = {'pr_auc': 'Train', 'val_pr_auc': 'Validation'}
    fig_auc.for_each_trace(lambda t: t.update(name = new_names_auc[t.name]))
    fig_auc.update_traces(line=dict(width=2.5))
    fig_auc.update_layout(
        width=700, height=500, autosize=False,
        template='plotly_white', hovermode='x unified', title_x=0.5
    )
    return fig_auc

def create_test_report_2(modelo, calibrated_threshold = 0.45):
    test_probs = (modelo.predict(X_test)).ravel()
    test_preds = test_probs > calibrated_threshold
    print("----------- TEST REPORT -----------")
    print(classification_report(y_test, test_preds))
    precision, recall, thresholds = precision_recall_curve(y_test, test_probs)
    pr_auc = auc(recall, precision)
    ap = average_precision_score(y_test, test_probs)
    print(f"PR-AUC Score: {pr_auc:.4f}")
    print(f"Average Precision (AP) Score: {ap:.4f}")
    return precision, recall, pr_auc, ap

# Función para graficar la curva Precision-Recall

def plot_PRAUC(precision, recall, pr_auc, ap, algorithm, name, colour):
    PRAUC_df = pl.DataFrame({"Recall": recall, "Precision": precision})
    light_PRAUC_df = PRAUC_df.gather_every(50)
    baseline = float(y_test.sum() / len(y_test))
    fig = px.line(
        light_PRAUC_df, 
        x='Recall', 
        y='Precision',
        title=f'<b>Precision-Recall Curve (PR-AUC): {name}</b>',
        labels={'Recall': '<b>Recall<b>', 'Precision': '<b>Precision<b>'}
    )
    fig.update_traces(
        line=dict(color=colour, width=4),
        name=f'{algorithm} (AUC = {pr_auc:.4f} | AP = {ap:.4f})',
        showlegend=True,
        hovertemplate='<b>Recall:</b> %{x:.4f}<br><b>Precision:</b> %{y:.4f}<extra></extra>'
    )
    fig.add_shape(
        type="line", x0=0, y0=baseline, x1=1, y1=baseline,
        line=dict(color="navy", width=2, dash="dash"),
    )
    fig.add_trace(px.scatter(x=[None], y=[None], labels={'x':'', 'y':''}).data[0])
    fig.data[-1].update(
        name=f'Random Classifier (Prop. = {baseline:.4f})', 
        showlegend=True, mode='lines',
        line=dict(color="navy", width=2, dash="dash")
    )
    fig.update_layout(
        width=700, height=500, title_x=0.5, autosize=False,
        xaxis=dict(range=[0.0, 1.0], gridcolor='rgba(0,0,0,0.1)'),
        yaxis=dict(range=[0.0, 1.05], gridcolor='rgba(0,0,0,0.1)'),
        plot_bgcolor='white',
        legend=dict(x=0.45, y=1.1, bgcolor='rgba(255, 255, 255, 0.8)')
    )
    return fig

# Función para calcular los costes de las predicciones del modelo
def calculate_financial_metrics(predictions_df):
    financial_metrics = (
        predictions_df
        .select(
            # Dinero Rescatado: Fraudes detectados (True Positives)
            Recovered_Amount = pl.when(pl.col("target") & pl.col("prediction")).then(pl.col("amount")).otherwise(0).sum(),
            # Dinero Perdido: Fraudes que el modelo no vio (False Negatives)
            Leaked_Amount = pl.when(pl.col("target") & ~pl.col("prediction")).then(pl.col("amount")).otherwise(0).sum(),
            # Dinero Bloqueado por Falsas Alarmas: (False Positives)
            False_Positive_Cost = pl.when(~pl.col("target") & pl.col("prediction")).then(pl.col("amount")).otherwise(0).sum(),
            # Total de dinero que estaba en riesgo de fraude
            Total_Fraud = pl.when(pl.col("target")).then(pl.col("amount")).otherwise(0).sum()
        )
        .with_columns(
            Recovery_Rate_Pct = pl.col("Recovered_Amount") / pl.col("Total_Fraud") * 100
        )
    )
    return financial_metrics
````

<br>A continuación calculamos el riesgo de cada tipo de comercio como diana de fraude. Para ello solo utilizamos las transacciones auditadas, y calculamos el factor de riesgo dividiendo el número de transacciones fradulentas entre el número de transacciones legítimas, para evitar problemas de escala, donde los valores que presentan una tasa entre 0 y 1 sean comparables a los valores mayor a 1, utilizaremos logaritmos, por ejemplo si un comercio presenta una tasa de 0.5 será el doble de segura que una con riesgo 1 riesgo, una tasa de 0.25 será 4 veces más segura que un comercio con tasa 1, es por ello que para escalar y hacer comparables estos valores utilizamos logaritmos, ya que:  
$\log (1/X) = -\log(X)$
Al valor obtenido lo denominaremos Log2RiskFactor

````python
mcc_description_df = (
    fraud_data_df
    .filter(pl.col("target").is_not_null())
    .pivot(
        on="target", index="mcc_description", 
        values="mcc_description", aggregate_function="len",
        on_columns=["false", "true"], 
    )
    .rename({"false": "Legitimate", "true": "Fraudulent"})
    .with_columns(
        mcc_rate = pl.col("Fraudulent")/(pl.col("Legitimate") + 1)
    )
    .with_columns(
        total_leg = pl.col("Legitimate").sum(),
        total_fraud = pl.col("Fraudulent").sum()
    )
    .with_columns(
        fraud_rate = pl.col("total_fraud")/pl.col("total_leg")
    )
    .with_columns(
        mcc_risk_factor = pl.col("mcc_rate")/pl.col("fraud_rate")
    )
    .with_columns(
        Log2_risk_factor = (pl.col("mcc_risk_factor") + 0.00001).log(base=2).round(2)
    )  
    .with_columns(
        mcc_risk_category = (
            pl.when(pl.col("Log2_risk_factor") > 2).then(pl.lit("High Risk"))
                .when(pl.col("Log2_risk_factor") < -2).then(pl.lit("Low Risk"))
                .otherwise(pl.lit("Baseline Risk"))
        )
    ) 
    .drop(["total_leg", "total_fraud", "fraud_rate", "mcc_rate"])
)

mcc_description_df.show(5)
````

En el siguiente gráfico representamos los valores obtenidos del Log2RiskFactor de cada comercio. Para discriminar los comercios con alto riesgo utilizamos un umbral superior a 2, representado con color rojo, y para bajo riesgo un umbral inferior a -2, representado con color azul, aquellos valores que fueron considerados neutros están representados en gris. Como podemos observar en el gráfico las compañías de cruceros y tiendas musciales son las que presentan mayor tasa de operaciones fraudulentas, mientras que los talleres automotrices y los clubes recreativos y deportivos son los más seguros.

````python
mcc_description_mem = mcc_description_df.collect()

color_risk = {
    'High Risk': '#d9534f',      
    'Baseline Risk': '#999999',  
    'Low Risk': '#337ab7'      
}

mcc_fig = px.scatter(
    mcc_description_mem,
    x="Legitimate",
    y="Fraudulent",
    color="mcc_risk_category",   
    color_discrete_map=color_risk,    
    hover_name="mcc_description", 
    hover_data=["Legitimate", "Fraudulent", "mcc_risk_factor", "Log2_risk_factor"],
    log_x=True,                   
    log_y=True,                  
    title="<b>Merchant Anomaly Detection and Risk Factor Analysis</b>",
    labels={"mcc_risk_category": "Risk Level", "mcc_risk_factor": "Risk Factor",
           "Log2_risk_factor": "Log2(Risk Factor)"},
    template="plotly_white"   
)


mcc_fig.update_traces(
    marker=dict(
        size=12,
        line=dict(width=1, color='Black'),
        opacity=0.8
    )
)

mcc_fig.update_layout(
    height=400, width=900, autosize=False,
    title_font_size=18, xaxis_title_font_size=12, yaxis_title_font_size=12,
    margin=dict(l=40, r=40, t=60, b=40),
    legend=dict(orientation="h", yanchor="top", y=1.07, xanchor="left", x=0.05)
)

mcc_fig.show()
````

````python
total_amountchip_df = (
    amountchip_fraud_FY2016_df
    .group_by("target", "use_chip")
    .agg(
        total_transactions = pl.col("amount").len(),
        total_amount = pl.col("amount").sum().round(2)
    )
    .with_columns(target = label_expr)
    .sort("total_amount", descending = True)
    .collect()
)

total_amountchip_df.show(10)
````

````python
#En esta sección definimos una función para graficar la cantidad defraudada y el total de transacciones
def total_chip_plot (df, y_total, y_axis):
    total_amountchip_plot = px.bar(
        df, 
        x="use_chip", 
        y=y_total, 
        color="use_chip",
        facet_row="target",             
        facet_row_spacing=0.1,         
        color_discrete_map=chip_color,
        category_orders={"use_chip": chip_order},
        text_auto=",",
        custom_data=["target"]         
    )
    
    total_amountchip_plot.update_layout(
        plot_bgcolor="white", paper_bgcolor="white", showlegend=True, 
        width=600, height=600, autosize=False,                     
        margin=dict(l=80, r=30, t=60, b=50),  
        title=dict(text="<b>Fiscal Year 2016 Transaction Breakdown by Payment Method</b>", x=0.5, font=dict(size=14, color="black")),  
        legend=dict(title="", orientation="v", yanchor="top", y=1.05, xanchor="left", x=0.8) 
    )
    
    total_amountchip_plot.update_traces(
        marker_line_color="#4A4A4A", marker_line_width=1.2,
        textposition="outside", textfont=dict(size=11, color="black"), width=0.5,
        hovertemplate="<b>%{x}</b><br>Status: %{customdata[0]}<br>Total: %{y:,}<extra></extra>"
    )
    
    total_amountchip_plot.update_yaxes(
        title= y_axis,
        gridcolor="#EAEAEA", showline=False, zeroline=False, 
        tickfont=dict(size=11, color="black"),
        matches=None, showticklabels=True
    )
    
    total_amountchip_plot.update_xaxes(
        title="", showgrid=False, linecolor="#D3D3D3", 
        tickfont=dict(size=11, color="black", weight="bold"),
        showticklabels=True
    )
    
    total_amountchip_plot.for_each_annotation(
        lambda a: a.update(
            text="<b>Legitimate</b>" if "Legitimate" in a.text 
            else ("<b>Fraudulent</b>" if "Fraudulent" in a.text 
            else f"<b>{a.text.split('=')[-1]}</b>")
        )
    )
    
    return total_amountchip_plot
````

En la siguiente figura, usando el código de colores establecido para el método de pago, están representadas las cantidades totales defraudadas y el dinero gastado en transacciones legítimas. Como se observa las transacciones online, son en las que mayor cantidad de dinera se defrauda, lo cual contrasta con la tendencia observada en las cantidades totales movidas en transferencias legítimas, donde las transacciones online son las que menor cuantía en transacciones representan.

Para hacer una comparación más adecuada entre los métodos de pago y una mayor tendecia al fraude, nosotros calcularemos la cantidad de dinero defraudada cada 10.000 transacciones por método de pago.

````python
ratio_fraud = (
    total_amountchip_df
    .pivot(
        on="target", index="use_chip", values="total_transactions", aggregate_function="sum"
    )
    .with_columns(
        Frauds_per_10k_Transactions =
            pl.col("Fraudulent") / (pl.col("Fraudulent") + pl.col("Legitimate")) * 10000
    )
    .sort("Frauds_per_10k_Transactions", descending = True)
)

ratio_fraud.show()
````

````python
location_fraud_unbiased_df = (
    fraud_data_df
    .with_columns(day = pl.col("date").cast(pl.Date))
    .filter((pl.col("target").is_not_null()))
    .with_columns(merchant_state = pl.col("merchant_state").fill_null("No_Country"))    
    .select("merchant_city", "merchant_state", "zip", 
            "merchant_latitude", "merchant_longitude",
            "client_latitude", "client_longitude",
            "target", "use_chip", "amount", "day", "date"
           )
)

Rows = location_fraud_unbiased_df.select(pl.len()).collect().item()
Columns = len(location_fraud_unbiased_df.collect_schema().names())

print(f"Rows: {Rows}, Columns: {Columns}")
location_fraud_unbiased_df.show(5)
````

````python
USA_states = [
    "CA", "NY", "TX", "FL", "ID", "CT", "IL", "PA", "OH", "GA", 
    "MI", "NC", "NJ", "VA", "WA", "AZ", "AK", "AL", "AR", "CO", 
    "DE", "HI", "IA", "IN", "KS", "KY", "LA", "MA", "MD", "ME", 
    "MN", "MO", "MS", "MT", "ND", "NE", "NH", "NM", "NV", "OK", 
    "OR", "RI", "SC", "SD", "TN", "UT", "VT", "WI", "WV", "WY",
    "DC"
]

location_fraud_unbiased_df = (
    location_fraud_unbiased_df
    .with_columns(
        merchant_country = pl.when(pl.col("merchant_state").is_in(USA_states))
                             .then(pl.lit("USA")).otherwise(pl.col("merchant_state"))
    )
)

location_fraud_unbiased_df.show(5)
````

````python
n_countries = location_fraud_unbiased_df.select(pl.col('merchant_country').n_unique()).collect().item()
print(f"Number of countries: {n_countries}")
````

Listamos los países que presentan como destino transacciones fraudulentas.

````python
fraudulent_countries_df = (
    location_fraud_unbiased_df
    .filter(pl.col("target"))
    .select(pl.col("merchant_country").unique())
    .collect() 
)

fraudulent_countries_list = fraudulent_countries_df["merchant_country"].to_list()
fraudulent_countries_list
````

A continuación haremos las transformaciones que conciernen al tratamiento de datos temporales creando funciones ventana e intervalos de confianza siguiendo la distribución de Von Mises, tal y como se recomienda en el artículo de 
http://dx.doi.org/10.1016/j.eswa.2015.12.030. Nosotros utilizaremos una ventana de 7 días. Normalizaremos los ángulos con los que trabajaremos y aplicamos transformaciones para el cálculo de la media periódica y de la desviación estándar periódica, cuyas fórmulas están indicadas en la metodología.

````python
temporal_features_fraud_data_df = (
    FE_fraud_data
    .select(["id", "client_id", "card_id", "date", "target"])
    .filter(unbiased_w7_expr)
    .with_columns(
        hour = pl.col("date").dt.hour(),
        minute = pl.col("date").dt.minute(),
    )
    .with_columns(
        hour_of_day = (pl.col("hour") + (pl.col("minute") / 60.0)),
        theta = normalize_angle((pl.col("hour") + (pl.col("minute") / 60.0)) * (2 * PI / 24.0))
    )
    .with_columns(
        sin_theta = pl.col("theta").sin(),
        cos_theta = pl.col("theta").cos(),
    )
    # Ordenamos los datos para hacer el rolling
    .sort(["client_id", "card_id", "date"])
    # En esta parte desplazamos los valores de la transacción previa para evitar el data leakage en el
    # cálculo de las métricas de la transacción actual
    .with_columns(
        sin_theta_prev = pl.col("sin_theta").shift(1).over(["client_id", "card_id"]).fill_null(0),
        cos_theta_prev = pl.col("cos_theta").shift(1).over(["client_id", "card_id"]).fill_null(0),
    )     
    # Seleccionamos una ventana de 7 días, como se recomienda en el artículo
    .rolling(index_column="date", period="7d", group_by=["client_id", "card_id"])
    .agg([
        pl.all().last(),
        (pl.len() -1).clip(lower_bound=1).alias("N_7d"),
        pl.col("sin_theta_prev").sum().alias("sum_sin_7d"),
        pl.col("cos_theta_prev").sum().alias("sum_cos_7d"),
    ])
    .filter(unbiased_expr)
    .with_columns(
        day_of_week = pl.col("date").dt.weekday(),
        # El weekend lo definiremos incluyendo los dias Viernes, Sábado y Domingo
        weekend = (pl.col("date").dt.weekday() >= 5),
        # La fórmula utilizada es la que se indica en el artículo
        mu_vM_7d = pl.arctan2("sum_sin_7d", "sum_cos_7d"),
        R_sq_7d = (pl.col("sum_sin_7d") / pl.col("N_7d")).pow(2) + (pl.col("sum_cos_7d") / pl.col("N_7d")).pow(2)
    )
    .with_columns(
        sigma_vM_7d = (1.0 / pl.col("R_sq_7d").clip(lower_bound=1e-6)).clip(lower_bound=1.0).log().sqrt()
    )
    .with_columns(
        SE_7D = pl.col("sigma_vM_7d") / pl.col("N_7d").sqrt()
    ) 
    .with_columns(
        me_90 = calculate_Zt_value("N_7d", 90) * pl.col("SE_7D"),
        me_95 = calculate_Zt_value("N_7d", 95) * pl.col("SE_7D"),
        me_99 = calculate_Zt_value("N_7d", 99) * pl.col("SE_7D"),
    )
    # Creamos los Intervalos de Confianza (90%, 95%, 99%) como se recomiendo en el articulo
    .with_columns(
        ic_lower_vm_90_7d = normalize_angle(pl.col("mu_vM_7d") - pl.col("me_90")),
        ic_upper_vm_90_7d = normalize_angle(pl.col("mu_vM_7d") + pl.col("me_90")),
        ic_lower_vm_95_7d = normalize_angle(pl.col("mu_vM_7d") - pl.col("me_95")),
        ic_upper_vm_95_7d = normalize_angle(pl.col("mu_vM_7d") + pl.col("me_95")),
        ic_lower_vm_99_7d = normalize_angle(pl.col("mu_vM_7d") - pl.col("me_99")),
        ic_upper_vm_99_7d = normalize_angle(pl.col("mu_vM_7d") + pl.col("me_99")),
    )
    # Creamos una etiqueta para indicar si la transacción se encuentra dentro los IC definidos antes
    .with_columns(
        is_time_in_IC_90 = 
            pl.when(pl.col("N_7d") == 1).then(True).otherwise(
                pl.when(pl.col("ic_lower_vm_90_7d") <= pl.col("ic_upper_vm_90_7d"))
                  .then(pl.col("theta").is_between(pl.col("ic_lower_vm_90_7d"), pl.col("ic_upper_vm_90_7d")))
                  .otherwise((pl.col("theta") >= pl.col("ic_lower_vm_90_7d")) | 
                             (pl.col("theta") <= pl.col("ic_upper_vm_90_7d"))
                            )
        ),
        is_time_in_IC_95 = 
            pl.when(pl.col("N_7d") == 1).then(True).otherwise(
                pl.when(pl.col("ic_lower_vm_95_7d") <= pl.col("ic_upper_vm_95_7d"))
                  .then(pl.col("theta").is_between(pl.col("ic_lower_vm_95_7d"), pl.col("ic_upper_vm_95_7d")))
                  .otherwise((pl.col("theta") >= pl.col("ic_lower_vm_95_7d")) | 
                             (pl.col("theta") <= pl.col("ic_upper_vm_95_7d"))
                            )
        ),
        is_time_in_IC_99 = 
            pl.when(pl.col("N_7d") == 1).then(True).otherwise(
                pl.when(pl.col("ic_lower_vm_99_7d") <= pl.col("ic_upper_vm_99_7d"))
                  .then(pl.col("theta").is_between(pl.col("ic_lower_vm_99_7d"), pl.col("ic_upper_vm_99_7d")))
                  .otherwise((pl.col("theta") >= pl.col("ic_lower_vm_99_7d")) | 
                             (pl.col("theta") <= pl.col("ic_upper_vm_99_7d"))
                            )
        ),
    )
    .drop(["target", "hour", "minute", "sin_theta_prev", "cos_theta_prev",
           "sum_sin_7d", "sum_cos_7d", "R_sq_7d", "SE_7D", "me_90", "me_95", "me_99"])
)

temporal_features_fraud_data_df.show(10)
````

En esta sección calcularemos la frecuencia con la que un cliente consume en un tipo de comercio, y la frecuencia de fraude de ese tipo de comercio, expresado con el parámetro Log2RiskFactor, descrito anteriormente. Estos cálculos lo haremos utilando un `.rolling` de 28 días. Dado que estos cálculos los haremos en función de si una transacción es legítima o fraudulenta, no aplicaremos una ventana de 28 días para filtrar previamente, ya que los valores nulos meterán ruido a los cálculos. Por tanto aplicamos directamente el filtro final de sesgo (`unbiased_expr`). 

````python
Metrics_OHE_fraud_data_df = (
    FE_fraud_data
    .select("client_id", "card_id", "date", "id", "target",
            "birth_year", "birth_month", "retirement_age", "acct_open_date", "expires",
            "errors", "card_brand", "card_type", "use_chip", "gender",
            "mcc_description",
    )
    .filter(unbiased_expr)
    # Para analizar el comportamiento de  compras crearemos un rolling de 28 días y calcularemos la frecuencia
    # de compra por tipo de comercio
    .rolling(index_column="date", period="28d", group_by=["client_id", "card_id", "mcc_description"])
    .agg([
        pl.all().last(),
        (pl.len() - 1).alias("mcc_count_prev")
    ])
     # Volvemos a ordenar nuestros datos para hacer un segundo rolling
    .sort(["client_id", "card_id", "date", "id"])
    .rolling(index_column="date", period="28d", group_by=["client_id", "card_id"])
    .agg([
        pl.all().last(),
        (pl.len() - 1).clip(lower_bound=1).alias("N_28d")
    ])
    .with_columns(
        mcc_rate = pl.col("mcc_count_prev") / pl.col("N_28d")
    )
    # En esta subsección calcularemos el Log2RiskFactor haciendo un rolling de 28 días
    # Volvemos a ordenar nuestros datos para hacer un tercer rolling ahora por fecha, ya que ahora 
    # no tendremos en cuenta el comportamiento del cliente sino el historial de todos los fraudes 
    .sort("date")
    .with_columns(
        Fraudulent = pl.when(pl.col("target")).then(1).otherwise(0),
        Legitimate = pl.when(~pl.col("target")).then(1).otherwise(0)
    )
    .with_columns(
        mcc_Fraudulent_prev = pl.col("Fraudulent").shift(1).over("mcc_description").fill_null(0),
        mcc_Legitimate_prev = pl.col("Legitimate").shift(1).over("mcc_description").fill_null(0),
    )
    .rolling(index_column="date", period="28d", group_by="mcc_description")
    .agg([
        pl.all().last(),
        pl.col("mcc_Fraudulent_prev").sum().fill_null(0).alias("mcc_fraud_prev"),
        pl.col("mcc_Legitimate_prev").sum().fill_null(0).alias("mcc_leg_prev"),
    ])
    .sort("date")
    .with_columns(
        Fraudulent_prev = pl.col("Fraudulent").shift(1).fill_null(0),
        Legitimate_prev = pl.col("Legitimate").shift(1).fill_null(0),
    )
    .rolling(index_column="date", period="28d")        
    .agg([
        pl.all().last(),
        pl.col("Fraudulent_prev").sum().fill_null(0).alias("total_fraud_prev"),
        pl.col("Legitimate_prev").sum().fill_null(0).alias("total_leg_prev")
    ])
    .with_columns(
        mcc_fraud_rate = pl.col("mcc_fraud_prev") / (pl.col("mcc_leg_prev") + 1),
        total_fraud_rate = pl.col("total_fraud_prev") / (pl.col("total_leg_prev") + 1)
    )
    .with_columns(
        mcc_risk_factor = pl.col("mcc_fraud_rate") / (pl.col("total_fraud_rate") + 0.00001)
    )
    .with_columns(
        Log2_risk_factor = (pl.col("mcc_risk_factor") + 0.00001).log(base=2).round(2)
    )
    
    # A continuación dejamos de trabajar con las métricas de los tipos de comercio y procedemos
    # a seguir con el resto de las variables
    .with_columns(
        pl.col("errors").cast(pl.String),
        pl.col("card_brand").cast(pl.String),
        pl.col("card_type").cast(pl.String),
        pl.col("use_chip").cast(pl.String),
        pl.col("gender").cast(pl.String),
        # En esta seccion nos dedicamos a calcular las últimas métricas temporales
        agemonth_at_event = (
            (pl.col("date").dt.year() - pl.col("birth_year")) * 12 +
            (pl.col("date").dt.month() - pl.col("birth_month"))
        ),
        days_acct_open = (pl.col("date") - pl.col("acct_open_date")).dt.total_days(),
        days_expires = (pl.col("expires") - pl.col("date")).dt.total_days()
    )
    .with_columns(
        retirement_month = pl.col("retirement_age") * 12 - pl.col("agemonth_at_event")    
    )
    # En esta seccion hacemos One Hot Encoding para el género, el tipo de transacción 
    # los errores, la marca y tipo de tarjeta
    .with_columns(
        No_Errors = pl.col("errors").is_null(),
        Bad_Card_Number = pl.col("errors").str.contains("Bad Card Number", literal=True).fill_null(False),
        Bad_CVV = pl.col("errors").str.contains("Bad CVV", literal=True).fill_null(False),
        Bad_Expiration = pl.col("errors").str.contains("Bad Expiration", literal=True).fill_null(False),
        Bad_PIN = pl.col("errors").str.contains("Bad PIN", literal=True).fill_null(False),
        Bad_Zipcode = pl.col("errors").str.contains("Bad Zipcode", literal=True).fill_null(False),
        Insufficient_Balance  = pl.col("errors").str.contains("Insufficient Balance ", literal=True).fill_null(False),
        Technical_Glitch = pl.col("errors").str.contains("Technical Glitch", literal=True).fill_null(False),

        Visa = pl.col("card_brand").str.contains("Visa", literal=True).fill_null(False),
        Discover = pl.col("card_brand").str.contains("Discover", literal=True).fill_null(False),
        Amex = pl.col("card_brand").str.contains("Amex", literal=True).fill_null(False),
        Mastercard = pl.col("card_brand").str.contains("Mastercard", literal=True).fill_null(False),

        Debit_Prepaid = pl.col("card_type").str.contains("Debit (Prepaid)", literal=True).fill_null(False),
        Credit = pl.col("card_type").str.contains("Credit", literal=True).fill_null(False),
        Debit = pl.col("card_type").str.contains("Debit", literal=True).fill_null(False),

        Online_Transaction = pl.col("use_chip").str.contains("Online Transaction", literal=True).fill_null(False),
        Chip_Transaction = pl.col("use_chip").str.contains("Chip Transaction", literal=True).fill_null(False),
        Swipe_Transaction = pl.col("use_chip").str.contains("Swipe Transaction", literal=True).fill_null(False),

        Male = pl.col("gender").str.contains("Male", literal=True),  
    )
    .sort(["client_id", "card_id", "date", "id"])        
    .drop([
        "errors", "card_brand", "card_type", "use_chip", "gender", "birth_year", "birth_month", "retirement_age",
        "acct_open_date", "expires", "mcc_description", 
        "N_28d", "mcc_risk_factor", "mcc_count_prev", "Fraudulent", "Legitimate", 
        "mcc_Fraudulent_prev", "mcc_Legitimate_prev", "mcc_fraud_prev", "mcc_leg_prev", 
        "Fraudulent_prev", "Legitimate_prev", "total_fraud_prev", "total_leg_prev", "mcc_fraud_rate", "total_fraud_rate", 
    ])    
)

Metrics_OHE_fraud_data_df.show(10)
````

### <span id="ML"></span> **3.8. Modelado por Machine Learning**

Materializamos nuestro lazyframe, ya que el algoritmo Random Forest solo acepta datos previamente materializados.

````python
df = features_df.collect()
````

Como se puede observar, existe un marcado desbalance en la variable objetivo (`target`) que se pretende clasificar mediante los modelos de aprendizaje automático. Por lo tanto para la evaluación del modelo haremos ciertos cambios como seleccionar solo una porción de las transacciones legítimas. 

````python
proportions = (
    df
    .group_by("target")
    .agg([
        pl.len().alias("cantidad"),
        (pl.len() / len(df) * 100).alias("porcentaje")
    ])
    .with_columns(target = label_expr)
)

proportions
````

Además debemos tener en cuenta que al haber un alto desbalance en las clases, en la evaluación del modelo, métricas como el `accuracy` no son las adecuadas para la selección del modelo más idóneos para nuestros objetivos, que es la detección de fraude, ya que si el modelo decide clasificar cualquier transacción como legítima, obtendrá un `accuracy` superior al 99%. Es por ello que las métricas que debemos  buscar que optimice el modelo son el `recall` (4), la `precision` (5) y el `F1 Score` (6).  
$$Precision = \frac{TP}{TP + FP}, \tag{4}$$
$$Recall = \frac{TP}{TP + FN}, \tag{5}$$
$$F1\text{-}Score = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}. \tag{6}$$

El conjunto de datos se particionó en tres subconjuntos: entrenamiento, validación y prueba, destinados respectivamente al ajuste de los modelos, la selección de hiperparámetros y la evaluación final de su rendimiento.

````python
False_df = df.filter(~pl.col("target"))
False_train_df = False_df.sample(fraction=0.70, seed=777)
False_valtest_df = False_df.join(False_train_df, on="id", how="anti")
False_val_df = False_valtest_df.sample(fraction=0.50, seed=777)
False_test_df = False_valtest_df.join(False_val_df, on="id", how="anti")

True_df = df.filter(pl.col("target"))
True_train_df = True_df.sample(fraction=0.70, seed=777)
True_valtest_df = True_df.join(True_train_df, on="id", how="anti")
True_val_df = True_valtest_df.sample(fraction=0.50, seed=777)
True_test_df = True_valtest_df.join(True_val_df, on="id", how="anti")

## Tomaremos solo 100_000 transaciones legitimas para el entrenamiento ya que hay un alto desbalance entre
## las transacciones legítimas y las fraudulentas
balanced_False_train_df = False_train_df.sample(n=100_000, seed=777)

train_df = pl.concat([balanced_False_train_df, True_train_df]).sample(fraction=1.0, shuffle=True, seed=777)
val_df = pl.concat([False_val_df, True_val_df]).sample(fraction=1.0, shuffle=True, seed=777)
test_df = pl.concat([False_test_df, True_test_df]).sample(fraction=1.0, shuffle=True, seed=777)

## Definimos los datos que usaremos para el entrenamiento, la validación, el testeo del modelo
## y la búsqueda de hiperparámetros
X_train = train_df.drop("target", "id")
y_train = train_df["target"].cast(pl.Int8).to_numpy()

X_val = val_df.drop("target", "id")
y_val = val_df["target"].cast(pl.Int8).to_numpy()

X_test = test_df.drop("target", "id")
y_test = test_df["target"].cast(pl.Int8).to_numpy()

X_search = pl.concat([X_train, X_val])
y_search = np.concatenate((y_train, y_val))

split_indices = np.zeros(len(X_search))
split_indices[:len(X_train)] = -1  
split_indices[len(X_train):] = 0   
pds = PredefinedSplit(test_fold=split_indices)
````

Como señalamos anteriormente existe un marcado desbalance en la variable objetivo (target) que se pretende clasificar mediante los modelos de aprendizaje automático.

````python
proportions = (
    df
    .group_by("target")
    .agg([
        pl.len().alias("cantidad"),
        (pl.len() / len(df) * 100).alias("porcentaje")
    ])
    .with_columns(target = label_expr)
)

proportions
````

El conjunto de datos se particionó en tres subconjuntos: entrenamiento, validación y prueba, destinados respectivamente al entrenamiento de los modelos, al seguimiento de su rendimiento durante el proceso de entrenamiento y a la evaluación final sobre datos no utilizados durante dicho proceso.

````python
False_df = df.filter(~pl.col("target"))
False_train_df = False_df.sample(fraction=0.70, seed=777)
False_valtest_df = False_df.join(False_train_df, on="id", how="anti")
False_val_df = False_valtest_df.sample(fraction=0.50, seed=777)
False_test_df = False_valtest_df.join(False_val_df, on="id", how="anti")

True_df = df.filter(pl.col("target"))
True_train_df = True_df.sample(fraction=0.70, seed=777)
True_valtest_df = True_df.join(True_train_df, on="id", how="anti")
True_val_df = True_valtest_df.sample(fraction=0.50, seed=777)
True_test_df = True_valtest_df.join(True_val_df, on="id", how="anti")

train_df = pl.concat([False_train_df, True_train_df]).sort(["client_id", "card_id", "date", "id"]) 
val_df = pl.concat([False_val_df, True_val_df]).sort(["client_id", "card_id", "date", "id"]) 
test_df = pl.concat([False_test_df, True_test_df]).sort(["client_id", "card_id", "date", "id"]) 

## Definimos los datos que usaremos para el entrenamiento, la validación y el testeo del modelo
scaler = StandardScaler()

X_train = scaler.fit_transform(
    train_df
    .drop("target", "client_id", "card_id", "date", "id")
    .cast(pl.Float32).to_numpy()
)
y_train = train_df["target"].cast(pl.Int8).to_numpy()

X_val = scaler.transform(
    val_df
    .drop("target", "client_id", "card_id", "date", "id")
    .cast(pl.Float32).to_numpy()
)
y_val = val_df["target"].cast(pl.Int8).to_numpy()

X_test = scaler.transform(
    test_df
    .drop("target", "client_id", "card_id", "date", "id")
    .cast(pl.Float32).to_numpy()
)
y_test = test_df["target"].cast(pl.Int8).to_numpy()

classes = np.unique(y_train)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
class_weight_dict = dict(zip(classes, weights))

pos = int(len(train_df)*proportions[1, 2]/100)
neg = len(train_df) - pos
initial_bias = np.log([pos / neg])
output_bias = tf.keras.initializers.Constant(initial_bias)
````

En esta sección se entrenará una red neuronal MLP y se evaluarán dos estrategias de función de pérdida para abordar el desequilibrio entre las clases:
- Binary Cross-Entropy + class_weight
- Binary Focal Cross-Entropy

````python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_pr_auc",
    mode="max",
    patience=10,
    restore_best_weights=True
)

def create_checkpoint(fpath):
    return tf.keras.callbacks.ModelCheckpoint(
        filepath=fpath,
        monitor="val_pr_auc",
        mode="max",
        save_best_only=True,
        verbose=0
)
````

````python
def create_model():
    return tf.keras.Sequential([
        layers.Input(shape=X_train.shape[1]),
    
        layers.Dense(512, activation="relu", kernel_regularizer=regularizers.l2(5e-4)),
        layers.BatchNormalization(),
        layers.Dropout(0.3),
    
        layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(5e-4)),
        layers.BatchNormalization(),
        layers.Dropout(0.3),
        
        layers.Dense(128, activation="relu", kernel_regularizer=regularizers.l2(5e-4)),
        layers.BatchNormalization(),
        layers.Dropout(0.2),
    
        layers.Dense(64, activation="relu", kernel_regularizer=regularizers.l2(5e-4)),
        layers.BatchNormalization(),
    
        layers.Dense(1, activation="sigmoid",  bias_initializer=output_bias)
    ])
````

En este primer modelo usaremos como función de pérdida `Binary Cross-Entropy + class_weight`

````python
model_mlp_1 = create_model()

model_mlp_1.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=[
        tf.keras.metrics.AUC(name='pr_auc', curve='PR'),
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall'),
    ]
)

checkpoint_1 = create_checkpoint(f"{path}/mldl/DL_MLP_1.keras")

history_mlp_1 = model_mlp_1.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=2048,
    validation_data=(X_val, y_val),
    class_weight=class_weight_dict,
    callbacks=[early_stopping, checkpoint_1],
    shuffle=True,
    verbose=0
)
````

A continuación, se representa la evolución de la métrica PR-AUC a lo largo de las épocas de entrenamiento, tanto para el conjunto de entrenamiento como para el conjunto de validación.

````python
WINDOW_SIZE = 5

df = df.sort("date")
total_rows = len(df)
train_index = int(total_rows * 0.70)
val_index = int(total_rows * 0.85)

VAL_CUTOFF_DATE = df["date"][train_index]
TEST_CUTOFF_DATE = df["date"][val_index]

train_df = (
    df.filter(pl.col("date") < VAL_CUTOFF_DATE)
    .sort(["client_id", "card_id", "date", "id"])
)
val_df = (
    df.filter((pl.col("date") >= VAL_CUTOFF_DATE) & (pl.col("date") < TEST_CUTOFF_DATE))
    .sort(["client_id", "card_id", "date", "id"])
)
test_df = (
    df.filter(pl.col("date") >= TEST_CUTOFF_DATE)
    .sort(["client_id", "card_id", "date", "id"])
)

excluded_columns = ["id", "client_id", "card_id", "date", "target", "window_target"]
columns_to_scale = train_df.select(pl.all().exclude(excluded_columns)).columns
scaler = StandardScaler()
scaler.fit(train_df.select(columns_to_scale).to_pandas())

def apply_scaling(df_polars, data_scaler, columns):
    transformed_df = pl.DataFrame(
        data_scaler.transform(df_polars.select(columns).to_pandas()), 
        schema=columns
    )
    return df_polars.with_columns([transformed_df[c] for c in columns])

def build_windows_by_id_join(df, window_size, is_train=True):            
    flat_ids_df = (
        df.select(["client_id", "card_id", "date", "id", "target"])
        .with_columns([
            *[pl.col("id").shift(window_size - 1 - i).over(["client_id", "card_id"]).alias(f"id_step_{i}") 
              for i in range(window_size)],
            pl.any_horizontal([
                pl.col("target").shift(i).over(["client_id", "card_id"]) 
                for i in range(window_size)
            ]).alias("window_target")
        ])
        .drop_nulls(subset=["id_step_0", "window_target"])
    )

    id_cols = [f"id_step_{i}" for i in range(window_size)]
    window_targets = flat_ids_df["window_target"].to_numpy().astype(np.int32)
    flat_ids_df = flat_ids_df.select(id_cols)

    feature_cols = df.select(pl.all().exclude(["client_id", "card_id", "date", "id", "target"])).columns
    raw_ids = df["id"].to_numpy()
    raw_features = df.select(feature_cols).to_numpy().astype(np.float32)
    del df
    gc.collect()
    
    id_to_index = {id_val: idx for idx, id_val in enumerate(raw_ids)}
    num_samples = len(flat_ids_df)
    num_features = len(feature_cols)
    X_3d = np.empty((num_samples, window_size, num_features), dtype=np.float32)
    
    for step in range(window_size):
        step_ids = flat_ids_df[f"id_step_{step}"].to_numpy()
        feature_indices = [id_to_index[id_val] for id_val in step_ids]
        X_3d[:, step, :] = raw_features[feature_indices]
        
    del raw_features, raw_ids, id_to_index, flat_ids_df
    gc.collect()
    
    return X_3d, window_targets
````

````python
train_df = apply_scaling(train_df, scaler, columns_to_scale)
X_train, y_train = build_windows_by_id_join(train_df, WINDOW_SIZE, is_train=True)
del train_df
gc.collect()
np.save(f"{path}/data/silver/Polars/X_train_w5.npy", X_train)
np.save(f"{path}/data/silver/Polars/y_train_w5.npy", y_train)
````

````python
val_df = apply_scaling(val_df, scaler, columns_to_scale)
X_val, y_val = build_windows_by_id_join(val_df, WINDOW_SIZE, is_train=False)
del val_df
gc.collect()
np.save(f"{path}/data/silver/Polars/X_val_w5.npy", X_val)
np.save(f"{path}/data/silver/Polars/y_val_w5.npy", y_val)
````

````python
test_df = apply_scaling(test_df, scaler, columns_to_scale)
X_test, y_test = build_windows_by_id_join(test_df, WINDOW_SIZE, is_train=False)
del test_df
gc.collect()
np.save(f"{path}/data/silver/Polars/X_test_w5.npy", X_test)
np.save(f"{path}/data/silver/Polars/y_test_w5.npy", y_test)
````

## <span id="discussion"></span> **4. Discusión**


Los resultados obtenidos muestran que las características demográficas, temporales, espaciales y relacionadas con el comportamiento de compra presentan diferencias entre las transacciones legítimas y fraudulentas. En primer lugar, se observaron diferencias en la distribución del fraude según la edad de los usuarios. La proporción de operaciones fraudulentas aumenta especialmente entre los usuarios de 75 años o más. Asimismo, se observó una elevada proporción de transacciones fraudulentas entre los usuarios de 18 años, edad mínima considerada en el conjunto de datos para la adquisición de una tarjeta bancaria, siendo este comportamiento más marcado en hombres. Estos resultados sugieren que la edad y, en menor medida, el sexo pueden estar asociados con diferentes patrones de exposición al fraude. No obstante, estas diferencias deben interpretarse con cautela, ya que una mayor frecuencia de fraude en un determinado grupo no implica necesariamente una mayor vulnerabilidad individual, pudiendo estar condicionada por la frecuencia y características de uso de las tarjetas en cada población.

En relación con el tipo de comercio, se observaron diferencias en la frecuencia de operaciones fraudulentas según la categoría del establecimiento. Las líneas de cruceros y las tiendas de música presentaron una mayor proporción relativa de operaciones fraudulentas, mientras que los establecimientos relacionados con el mantenimiento de vehículos y los servicios deportivos y recreativos mostraron una menor incidencia. Estas diferencias ponen de manifiesto la relevancia de incorporar información relacionada con el comercio en los modelos de detección, ya que determinadas categorías pueden presentar patrones de riesgo diferenciados. Sin embargo, estos resultados deben entenderse como una asociación observada en el conjunto de datos analizado y no como una medida general del riesgo real de cada sector comercial.

Una de las principales características del conjunto de datos fue la presencia de un elevado número de transacciones sin auditoría. Aproximadamente el 33 % de las operaciones carecía de una etiqueta que permitiera determinar si se trataba de una transacción legítima o fraudulenta. Esta circunstancia constituye una limitación relevante para el análisis supervisado, dado que estas observaciones no pueden utilizarse directamente para evaluar el rendimiento de los modelos. Por este motivo, las transacciones no auditadas fueron excluidas del análisis de las variables relacionadas con la condición de fraude y del entrenamiento de los modelos. Sin embargo, se conservaron para la generación de características relacionadas con el comportamiento histórico de los clientes que no requieren conocer previamente si una operación es fraudulenta o legítima. De esta forma, fue posible aprovechar parte de la información disponible sin introducir directamente la etiqueta desconocida en el proceso de aprendizaje.

Tras la exclusión de las transacciones no auditadas, se observó un marcado desequilibrio de clases, con únicamente un 0,29 % de operaciones fraudulentas frente al 99,71 % de transacciones legítimas. Este desequilibrio constituye una de las principales dificultades del problema, ya que un modelo podría alcanzar una elevada exactitud clasificando la práctica totalidad de las operaciones como legítimas y, al mismo tiempo, presentar una capacidad de detección de fraude insuficiente. Por ello, el desequilibrio fue considerado tanto durante el entrenamiento mediante estrategias de ponderación de las clases como durante la evaluación, priorizando métricas adecuadas para escenarios de clasificación altamente desbalanceados, como Precision-Recall y PR-AUC.

Desde una perspectiva temporal, los resultados muestran una mayor frecuencia de operaciones fraudulentas durante el mes de marzo y una concentración relativa en viernes y domingos. Asimismo, el intervalo comprendido entre las 10:00 y las 12:00 horas presentó una mayor frecuencia de operaciones fraudulentas. También se identificó una periodicidad aproximada de siete días, junto con una concentración de determinadas operaciones hacia el final de mes. La presencia de estos patrones temporales respalda la incorporación de características capaces de representar la periodicidad y el comportamiento histórico de las transacciones. En este sentido, la información temporal no solo permite caracterizar cuándo se producen las operaciones, sino también identificar desviaciones respecto al comportamiento habitual de un determinado cliente.

Otro resultado relevante fue la concentración del fraude en las transacciones realizadas en línea. Estas operaciones no solo concentraron una proporción importante de las transacciones fraudulentas, sino también el mayor volumen económico asociado al fraude. Además, las distribuciones de los importes de las operaciones fraudulentas mostraron valores generalmente superiores a los observados en las operaciones legítimas, incluyendo valores extremos más elevados. Este comportamiento se mantuvo al analizar los diferentes métodos de pago. Estos resultados sugieren que el importe de la transacción puede proporcionar información relevante para la detección, especialmente cuando se analiza en relación con el comportamiento histórico del cliente. Una cantidad elevada no constituye por sí misma una evidencia de fraude, pero una operación cuyo importe se desvía significativamente del patrón habitual de gasto del usuario puede aportar información discriminante al modelo.

En cuanto a la dimensión geográfica, Italia concentró la mayor frecuencia de transacciones fraudulentas realizadas en el extranjero dentro del conjunto de datos, mientras que las transacciones fraudulentas realizadas localmente se concentraron principalmente en Ohio. Dado que los usuarios del conjunto de datos son residentes de Estados Unidos, estas diferencias permiten observar la posible relevancia de la localización geográfica como elemento contextual de una transacción. No obstante, al igual que ocurre con las diferencias entre categorías comerciales, estos resultados dependen de la distribución geográfica específica del conjunto de datos y no deben interpretarse directamente como una medida general del riesgo de fraude de un país o región.

La incorporación de información histórica permitió, además, observar patrones de comportamiento que no serían identificables mediante el análisis aislado de cada transacción. El código desarrollado para analizar el comportamiento de compra de los clientes permitió identificar, en determinados casos, operaciones atípicas asociadas con transacciones fraudulentas. Entre estos comportamientos se observaron ráfagas de operaciones concentradas en periodos de tiempo reducidos, realizadas en horarios poco habituales, con importes superiores a los habituales del cliente o en categorías de comercios poco frecuentadas previamente. En el análisis de las transacciones para las que se disponía de información de localización, se observó además que la primera operación fraudulenta podía producirse en un establecimiento previamente utilizado por el cliente, mientras que las operaciones fraudulentas posteriores se asociaban con establecimientos que no formaban parte de su historial habitual de compras. Este resultado resulta especialmente relevante, ya que muestra que el fraude no necesariamente comienza con una transacción completamente anómala, sino que puede presentar una evolución observable a través de una secuencia de operaciones.

Estas observaciones sirvieron de base para la generación de las características utilizadas posteriormente en los modelos predictivos. Se incorporaron características temporales destinadas a representar la periodicidad y el comportamiento histórico de las transacciones, incluyendo variables basadas en la distribución de von Mises, así como características relacionadas con la cuantía de los desembolsos. También se generaron características espaciales basadas en la distancia entre localizaciones, calculada mediante la fórmula de Haversine, y variables destinadas a representar el riesgo asociado a las diferentes categorías de comercio. De este modo, el proceso de ingeniería de características permitió trasladar al modelo parte de los patrones identificados durante el análisis exploratorio del comportamiento de los clientes. La información de localización se utilizó para caracterizar el comportamiento espacial, evitando utilizar directamente la condición de fraude como fuente de información para estas características.

Finalmente, la comparación de los modelos desarrollados mostró que LightGBM presentó el mejor rendimiento entre los modelos evaluados cuando se optimizó utilizando F1 como criterio de *scoring*. Este modelo alcanzó un valor de PR-AUC de 0,7463 sobre el conjunto de evaluación. Este resultado indica una capacidad relevante para discriminar las transacciones fraudulentas en un escenario caracterizado por un fuerte desequilibrio de clases. La utilización de PR-AUC resulta especialmente apropiada en este contexto, ya que permite evaluar conjuntamente la capacidad del modelo para recuperar operaciones fraudulentas y mantener una precisión adecuada ante la elevada proporción de transacciones legítimas. No obstante, el rendimiento obtenido también pone de manifiesto que existe margen de mejora, especialmente teniendo en cuenta la naturaleza dinámica del fraude y la dificultad de identificar patrones fraudulentos que pueden cambiar con el tiempo.
